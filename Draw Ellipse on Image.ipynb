```python
import cv2
```

```python
old_img=cv2.imread("picture/fb16.jpg")
old_img=cv2.resize(old_img,(500,600))

new_img=cv2.ellipse(img=old_img,center=(250,160),
                    axes=(140,100), # For example: Major axis 140 pixels and minor axis 100 pixels from the center.
                    angle=80,startAngle=36,endAngle=360,color=(0,255,255),thickness=3,lineType=8)

cv2.imshow("image",new_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
