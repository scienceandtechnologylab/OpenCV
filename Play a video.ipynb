```python
import cv2
```

```python
cap=cv2.VideoCapture("video/song1.mp4")

while cap.isOpened():
    ret,frame=cap.read()
    if ret==True:
        #frame=cv2.resize(frame,(500,400))
        cv2.imshow("video",frame)
        if cv2.waitKey(25) & 0xff==ord("q"):
            break
    else:
        break

cap.release()
cv2.destroyAllWindows()
```
