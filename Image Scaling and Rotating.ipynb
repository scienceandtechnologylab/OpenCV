```python
import cv2
```

# Scaling

```python
img=cv2.imread("picture/girl_21.jpg")

# 1. Scaling directly to a specific pixel size (Width: 450, Height: 500)
img1=cv2.resize(src=img,dsize=(450,500),interpolation=cv2.INTER_CUBIC)

# 2. Reducing by 50% using scaling factors (fx=0.5, fy=0.5)
half_scaled_img = cv2.resize(img, (0, 0), fx=0.5, fy=0.5)

cv2.imshow("Resize image",img1)
cv2.imshow("Image",img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

# Rotation

```python
img=cv2.imread("picture/girl_10.jpg")
img=cv2.resize(img,(450,500))

w,h=img.shape[0],img.shape[1]
m=cv2.getRotationMatrix2D(center=(w/2,h/2),angle=40,scale=1)
print(m)
new_img=cv2.warpAffine(src=img,M=m,dsize=(w,h))

cv2.imshow("Original Image",img)
cv2.imshow("Rotate image",new_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
