```python
import cv2
```

```python
cap=cv2.VideoCapture("video/song1.mp4")
c=0
while True:
    ret,frame=cap.read()

    if ret==True:
        #frame=cv2.resize(frame,(700,500))
        filename="video-image//img_"+str(c)+".png"
        cv2.imwrite(filename,frame)
        
        cv2.imshow("Original Video",frame)
        c=c+1
        if cv2.waitKey(25) & 0xff ==ord('q'):
            break

    else:
        break
cap.release()
cv2.destroyAllWindows()
```
