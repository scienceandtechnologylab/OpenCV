```python
import cv2
import numpy as np
```

```python
# Create white image
new_img=np.ones((500,500,3),np.uint8)*255
cv2.imshow("white image",new_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Create black image
new_img=np.zeros((500,500,3),np.uint8)
cv2.imshow("Black image",new_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
