```python
import cv2
import numpy as np
```

```python
# একটি খালি/ডামি কলব্যাক ফাংশন
def change(r):
    pass
```

```python
img=np.zeros((500,500,3),np.uint8)  #black image

cv2.namedWindow("color")

# Positional Argument ব্যবহার করে ট্র্যাকার তৈরি করা হয়েছে
cv2.createTrackbar("R", "color", 0, 255, change)
cv2.createTrackbar("G", "color", 0, 255, change)
cv2.createTrackbar("B", "color", 0, 255, change)

while True:
    cv2.imshow("color",img)
    if cv2.waitKey(1) & 0xff==ord("q"):
        break
    # Positional Argument ব্যবহার করে মান নেওয়া হচ্ছে
    r = cv2.getTrackbarPos("R", "color")
    g = cv2.getTrackbarPos("G", "color")
    b = cv2.getTrackbarPos("B", "color")

    # OpenCV BGR ফরম্যাট ব্যবহার করে
    img[:]=[b,g,r]

cv2.destroyAllWindows()
```
