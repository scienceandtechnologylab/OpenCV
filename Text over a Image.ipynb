```python
import cv2
```

```python
# All fontFase value
cv2.FONT_HERSHEY_COMPLEX=3
cv2.FONT_HERSHEY_COMPLEX_SMALL=5
cv2.FONT_HERSHEY_DUPLEX=3
cv2.FONT_HERSHEY_PLAIN=1
cv2.FONT_HERSHEY_SCRIPT_COMPLEX=7
cv2.FONT_HERSHEY_SCRIPT_SIMPLEX=6
cv2.FONT_HERSHEY_SIMPLEX=0
cv2.FONT_HERSHEY_TRIPLEX=4
cv2.FONT_ITALIC

# All lineType value
cv2.FILLED=-1
cv2.LINE_4=4
cv2.LINE_8=8
cv2.LINE_AA=16
```

```python
img=cv2.imread("picture/fb16.jpg")

txt_img=cv2.putText(img=img,
                    text="Tamannah",
                    org=(200,50),
                    fontFace=cv2.FONT_HERSHEY_COMPLEX,
                    fontScale=2,   # Font size
                    color=(0,0,255),  #text color
                    thickness=3,
                    lineType=cv2.LINE_8,
                    bottomLeftOrigin=False)

cv2.imshow("Image",txt_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
