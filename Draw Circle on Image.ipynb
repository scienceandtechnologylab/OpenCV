```python
import cv2
```

```python
old_img=cv2.imread("picture/fb16.jpg")
old_img=cv2.resize(old_img,(500,600))

txt=cv2.putText(old_img,text="Newton",org=(10,50),color=(255,0,255),
                fontFace=cv2.FONT_HERSHEY_DUPLEX,fontScale=3,thickness=3,lineType=8,bottomLeftOrigin=False)

new_img=cv2.circle(img=txt,center=(250,160),radius=120,color=(0,255,255),thickness=3,lineType=8)

cv2.imshow("image",new_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
#if you want to hide face you can use thickness=-1 value
old_img=cv2.imread("picture/fb16.jpg")
old_img=cv2.resize(old_img,(500,600))

new_img=cv2.circle(img=old_img,center=(250,160),radius=120,color=(0,255,255),thickness=-1,lineType=8)

cv2.imshow("image",new_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
