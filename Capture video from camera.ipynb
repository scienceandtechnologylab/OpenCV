```python
import cv2
```

```python
cap=cv2.VideoCapture(0)

while True:
    ret,frame=cap.read()
    if ret==True:
        frame=cv2.resize(frame,(500,400))
        cv2.imshow("My video",frame)
        if cv2.waitKey(25) & 0xff==ord("q"):
            break
    else:
        break
cap.release()
cv2.destroyAllWindows()
```
