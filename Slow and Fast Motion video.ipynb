```python
import cv2
```

# Normal Motion

```python
#Normal motion
cap=cv2.VideoCapture("video/song1.mp4")

while cap.isOpened():
    ret,frame=cap.read()
    if ret==True:
        cv2.imshow("video",frame)
        if cv2.waitKey(25) & 0xff==ord("q"):
            break
    else:
        break

cap.release()
cv2.destroyAllWindows()
```
